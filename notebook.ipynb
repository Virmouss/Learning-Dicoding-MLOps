# Sentiment Classification - Game Reviews / Steam Reviews

```python
import os
import sys
from typing import Text
 
from absl import logging
from tfx.orchestration import metadata, pipeline
from tfx.orchestration.beam.beam_dag_runner import BeamDagRunner
```

```python
PIPELINE_NAME = "izzan_virm-pipeline"
 
# pipeline inputs
DATA_ROOT = "Data"
TRANSFORM_MODULE_FILE = "review_transform.py"
TRAINER_MODULE_FILE = "review_trainer.py"
TUNER_MODULE_FILE = "review_tuner.py"
# requirement_file = os.path.join(root, "requirements.txt")
 
# pipeline outputs
OUTPUT_BASE = "output"
serving_model_dir = os.path.join(OUTPUT_BASE, 'serving_model')
pipeline_root = os.path.join(OUTPUT_BASE, PIPELINE_NAME)
metadata_path = os.path.join(pipeline_root, "metadata.sqlite")
```

```python
def init_local_pipeline(
        components, pipeline_root: Text
        ) -> pipeline.Pipeline: # type: ignore
    
    logging.info(f"Pipeline root set to: {pipeline_root}")
    beam_args = [
        "--direct_running_mode=multi_processing"
        # 0 auto-detect based on on the number of CPUs available 
        # during execution time.
        "----direct_num_workers=0" 
    ]
    
    return pipeline.Pipeline(
        pipeline_name=PIPELINE_NAME,
        pipeline_root=pipeline_root,
        components=components,
        enable_cache=True,
        metadata_connection_config=metadata.sqlite_metadata_connection_config(
            metadata_path
        ),
        eam_pipeline_args=beam_args
    )
```

```python
from component import init_components

logging.set_verbosity(logging.INFO)
components = init_components(
    DATA_ROOT,
    training_module=TRAINER_MODULE_FILE,
    tuner_module=TUNER_MODULE_FILE,
    transform_module=TRANSFORM_MODULE_FILE,
    # training_steps=5000,
    # eval_steps=1000,
    serving_model_dir=serving_model_dir,
)
    
pipeline = init_local_pipeline(components, pipeline_root)
BeamDagRunner().run(pipeline=pipeline)
```
